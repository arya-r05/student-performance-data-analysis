```python
import pandas as pd

df = pd.read_csv("StudentsPerformance.csv")

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df[['math score', 'reading score', 'writing score']].mean()
```

```python
df.groupby("gender")[["math score", "reading score", "writing score"]].mean()
```

```python
import matplotlib.pyplot as plt

df.groupby("gender")["math score"].mean().plot(kind="bar")
plt.title("Average Math Score by Gender")
plt.ylabel("Average Score")
plt.show()
```

```python
df.groupby("gender")["reading score"].mean().plot(kind="bar")
plt.title("Average Reading Score by Gender")
plt.ylabel("Average Score")
plt.show()
```

```python
df.groupby("gender")["writing score"].mean().plot(kind="bar")
plt.title("Average Writing Score by Gender")
plt.ylabel("Average Score")
plt.show()
```

```python
df.groupby("test preparation course")[["math score", "reading score", "writing score"]].mean()
```

```python
df.groupby("lunch")[["math score", "reading score", "writing score"]].mean()
```

This project analyzed student performance data using Python, Pandas, and Matplotlib.

The analysis revealed that:

Female students perform better in reading and writing.

Male students perform slightly better in mathematics.

Completing the test preparation course significantly improves performance.

Students with standard lunch tend to score higher than those with free/reduced lunch.

These insights show how socio-economic and preparatory factors influence academic outcomes.
